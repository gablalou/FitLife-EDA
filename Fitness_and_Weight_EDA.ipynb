<span style="font-size: 24px;">2. Participant Demographics Analysis</span>

### Demographics Analysis

```python
def analyze_demographics():
    # Create figure for demographics
    fig = plt.figure(figsize=(20,6))
    
    # Age Distribution
    age_count = df['age'].value_counts().reset_index()
    age_count.columns = ['age', 'count']
    fig_bar = px.bar(age_count, x="age", y='count', color="age", barmode="group")
    fig_bar.update_layout(title="Age Distribution by Count", title_x=0.5)
    fig_bar.show()
    
    # Gender Distribution
    gender_count = df['gender'].value_counts().reset_index()
    gender_count.columns = ['gender', 'percentage']
    fig_gender = px.pie(gender_count, values='percentage', names='gender', 
                 title='Gender Distribution')
    fig_gender.show()
    
    # BMI Distribution by Gender
    fig_bmi = px.box(df, x="gender", y="bmi", color="gender", notched=True)
    fig_bmi.update_layout(title="BMI Distribution by Gender")
    fig_bmi.show()
    
analyze_demographics()    
```

<span style="font-size: 24px;">3. Activity Analysis</span>

```python
print("\n Activity Patterns")
print("-" * 50)
def analyze_activities():
    # 1. Activity Distribution (Bar Chart)
    activity_counts = df['activity_type'].value_counts().reset_index()
    activity_counts.columns = ['activity_type', 'count']
    
    fig_activity_dist = px.bar(
        activity_counts, 
        x='count', 
        y='activity_type', 
        orientation='h', 
        color='activity_type', 
        title='Activity Distribution',
        color_discrete_sequence=px.colors.sequential.Viridis
    )
    fig_activity_dist.update_layout(xaxis_title='Number of Sessions', yaxis_title='Activity Type')
    fig_activity_dist.show()
    
    # 2. Activity Intensity Heatmap
    activity_intensity = pd.crosstab(df['activity_type'], df['intensity'])
    
    fig_intensity_heatmap = go.Figure(
        data=go.Heatmap(
            z=activity_intensity.values,
            x=activity_intensity.columns,
            y=activity_intensity.index,
            colorscale='YlOrRd'
        )
    )
    fig_intensity_heatmap.update_layout(
        title='Activity Intensity Heatmap',
        xaxis_title='Intensity Level',
        yaxis_title='Activity Type'
    )
    fig_intensity_heatmap.show()
    
    # 3. Average Calories Burned by Activity (Bar Chart)
    avg_calories = df.groupby('activity_type')['calories_burned'].mean().sort_values(ascending=False).reset_index()
    
    fig_avg_calories = px.bar(
        avg_calories, 
        x='calories_burned', 
        y='activity_type', 
        orientation='h', 
        color='activity_type', 
        title='Average Calories Burned per Activity',
        color_discrete_sequence=px.colors.sequential.Inferno
    )
    fig_avg_calories.update_layout(xaxis_title='Average Calories Burned', yaxis_title='Activity Type')
    fig_avg_calories.show()

# Call the function to display the plots
analyze_activities()
```

<span style="font-size: 24px;">4. Health Metrics Analysis</span>

```python
print("\n Health Metrics")
print("-" * 50)

def analyze_health_metrics():
    health_metrics = ['bmi', 'avg_heart_rate', 'hours_sleep', 'stress_level', 'daily_steps',
                      'hydration_level', 'calories_burned']
    # Analyze correlation between health metrics
    plt.figure(figsize=(12,8))
    correlation = df[health_metrics].corr()
    sns.heatmap(correlation, annot=True, cmap='rocket', center=0)
    plt.title('Correlation Between Health Metrics')
    plt.tight_layout()
    plt.show()
    
    # Analyze Distribution of Health Conditions
    health_conditions_count = df['health_condition'].value_counts().reset_index()
    health_conditions_count.columns = ['health_condition', 'count']
    fig_bar = px.bar(health_conditions_count, x='health_condition', y='count', color="health_condition", barmode="group")
    fig_bar.update_layout(title='Distribution of Health Conditions')
    fig_bar.show()
    
analyze_health_metrics()
```

<span style="font-size: 24px;">5. Time Series Analysis</span>

```python
print("\n Time Series Patterns")
print("-" * 50)

def analyze_time_patterns():
    # 1. Activity Frequency Over Time
    daily_activities = df.groupby('date')['activity_type'].count().reset_index()
    daily_activities.columns = ['date', 'activity_count']

    fig_daily = px.line(
        daily_activities, 
        x='date', 
        y='activity_count', 
        title='Activity Frequency Over Time',
        labels={'date': 'Date', 'activity_count': 'Number of Activities'},
        line_shape='spline'
    )
    fig_daily.update_layout(xaxis=dict(tickangle=45))
    fig_daily.show()

    # 2. Weekly Patterns
    df['weekday'] = df['date'].dt.day_name()
    weekday_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
    weekday_activities = df['weekday'].value_counts().reindex(weekday_order).reset_index()
    weekday_activities.columns = ['weekday', 'activity_count']

    fig_weekly = px.bar(
        weekday_activities, 
        x='weekday', 
        y='activity_count', 
        title='Activity Distribution by Day of the Week',
        labels={'weekday': 'Day of the Week', 'activity_count': 'Number of Activities'},
        color='weekday', 
        category_orders={'weekday': weekday_order},
        color_discrete_sequence=px.colors.qualitative.Set2
    )
    fig_weekly.update_layout(xaxis=dict(tickangle=45))
    fig_weekly.show()

analyze_time_patterns()
```

<span style="font-size: 24px;">6. Fitness and Weight Progress</span>

```python
print("\nFitness Progress Analysis")
print("-" * 50)

def analyze_fitness_progress():
    # Average Weight Change Over Time
    monthly_weight = df.groupby([df['date'].dt.to_period('M'), 'gender'])['weight_kg'].mean().reset_index()
    monthly_weight['date'] = monthly_weight['date'].astype(str)
    
    plt.figure(figsize=(15, 6))
    for gender in monthly_weight['gender'].unique():
        gender_data = monthly_weight[monthly_weight['gender'] == gender]
        plt.plot(gender_data['date'], gender_data['weight_kg'], label=gender, marker='o')
    
    plt.title('Average Weight Progress by Gender')
    plt.xlabel('Month')
    plt.ylabel('Average Weight (kg)')
    plt.legend()
    plt.xticks(rotation=45)
    plt.show()
    
    # Fitness Level Distribution
    plt.figure(figsize=(12, 6))
    sns.violinplot(data=df, x='health_condition', y='fitness_level')
    plt.title('Fitness Level Distribution by Health Condition')
    plt.xticks(rotation=45)
    plt.show()

analyze_fitness_progress()
```

<span style="font-size: 24px;">7. Key Insights</span>

```python
print("\nKey Dataset Insights")
print("-" * 50)

def generate_insights():
    insights = {
        'total_participants': df['participant_id'].nunique(),
        'total_activities': len(df),
        'avg_activities_per_person': len(df) / df['participant_id'].nunique(),
        'most_popular_activity': df['activity_type'].mode().iloc[0],
        'avg_calories_per_session': df['calories_burned'].mean(),
        'avg_daily_steps': df['daily_steps'].mean(),
        'avg_sleep_hours': df['hours_sleep'].mean()
    }
    
    print("\nKey Insights from the Dataset:")
    print(f"Total Participants: {insights['total_participants']:,}")
    print(f"Total Activities Recorded: {insights['total_activities']:,}")
    print(f"Average Activities per Person: {insights['avg_activities_per_person']:.1f}")
    print(f"Most Popular Activity: {insights['most_popular_activity']}")
    print(f"Average Calories Burned per Session: {insights['avg_calories_per_session']:.1f}")
    print(f"Average Daily Steps: {insights['avg_daily_steps']:,.0f}")
    print(f"Average Sleep Hours: {insights['avg_sleep_hours']:.1f}")

generate_insights()
```
